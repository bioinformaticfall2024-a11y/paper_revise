## Graph-only + InfoNCE control

This notebook defines the requested control loss and runs a **synthetic smoke test only**. It does **not** contain an observed DAVIS/KIBA experimental result. The actual control requires retraining with the exact data split, seeds, and training configuration.

```python
import torch
import torch.nn.functional as F
def symmetric_info_nce(a,b,tau=0.07):
    a=F.normalize(a,dim=-1); b=F.normalize(b,dim=-1)
    logits=(a@b.T)/tau
    labels=torch.arange(a.size(0),device=a.device)
    return F.cross_entropy(logits,labels)+F.cross_entropy(logits.T,labels)
def total_loss(pred,y,z_ligand,z_protein,alpha=0.5,tau=0.07):
    mse=F.mse_loss(pred.view_as(y),y); nce=symmetric_info_nce(z_ligand,z_protein,tau)
    return mse+alpha*nce,mse.detach(),nce.detach()
torch.manual_seed(7)
pred=torch.randn(8); y=torch.randn(8); zl=torch.randn(8,32); zp=torch.randn(8,32)
loss,mse,nce=total_loss(pred,y,zl,zp)
print("Synthetic smoke test only")
print("total_loss =",float(loss))
print("mse =",float(mse))
print("InfoNCE =",float(nce))
```
