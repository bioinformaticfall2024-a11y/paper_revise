## Radar-plot MSE normalization

The manuscript should state one transform: min-max normalize MSE, then invert as `1 - normalized_MSE`.

```python
def radar_mse_score(mse,mse_min,mse_max):
    if mse_max<=mse_min: raise ValueError('mse_max must be > mse_min')
    return 1.0-(mse-mse_min)/(mse_max-mse_min)
example=[0.126,0.130,0.145,0.194]
mn,mx=min(example),max(example)
for x in example:
    print(f'MSE {x:.3f} -> radar score {radar_mse_score(x,mn,mx):.4f}')
```
