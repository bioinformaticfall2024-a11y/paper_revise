## DGraphDTA KIBA baseline correction

This notebook recalculates the descriptive differences after replacing the incorrect KIBA DGraphDTA row with the value reported in the primary DGraphDTA paper. **These are cross-study descriptive differences, not matched head-to-head effects.**

```python
baseline = {"model": "DGraphDTA (published)", "MSE": 0.126, "CI": 0.904, "Pearson": 0.903}
models = [
    {"model": "FGgraphDTA", "MSE": 0.145, "CI": 0.865, "Pearson": 0.873},
    {"model": "PVgraphDTA", "MSE": 0.130, "CI": 0.898, "Pearson": 0.877},
]

rows=[]
for m in models:
    rows.append({
        'Model': m['model'],
        'MSE': m['MSE'],
        'MSE descriptive difference vs published DGraphDTA (%)': (m['MSE']-baseline['MSE'])/baseline['MSE']*100,
        'CI delta': m['CI']-baseline['CI'],
        'Pearson delta': m['Pearson']-baseline['Pearson'],
    })
rows
```

```python
for r in rows:
    print(f"{r['Model']}: MSE={r['MSE']:.3f}; MSE difference={r['MSE descriptive difference vs published DGraphDTA (%)']:+.2f}%; CI delta={r['CI delta']:+.3f}; Pearson delta={r['Pearson delta']:+.3f}")
```
